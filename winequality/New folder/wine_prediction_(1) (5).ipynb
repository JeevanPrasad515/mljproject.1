```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

```python
from sklearn.preprocessing import MinMaxScaler
scl = MinMaxScaler().fit(X_train)
X_train_norm = scl.transform(X_train)
X_test_norm = scl.transform(X_test)
```

```python
# from PLS import PLS
from sklearn.svm import SVC
from sklearn.svm import LinearSVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import ExtraTreesClassifier
from xgboost import XGBClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from lightgbm import LGBMClassifier
```

# **1. DecisionTree**

```python
from sklearn.model_selection import cross_val_predict
model = DecisionTreeClassifier()
y_predict = cross_val_predict(model, X_train_norm, y_train, cv=10)
dt_train_acc = sum(y_train==y_predict)/len(y_train) #train accuracy
print("Train Accuracy = " + str(dt_train_acc))
```

```python
model = DecisionTreeClassifier()
model.fit(X_train_norm, y_train)
y_predict = model.predict(X_test_norm)
dt_test_acc = sum(y_test == y_predict)/len(y_test) #test accuracy
print("Test Accuracy = " + str(dt_test_acc))
```

# **2. RandomForest**

```python
from sklearn.model_selection import cross_val_predict
param = [20, 50, 100, 200]
allyp = [cross_val_predict(RandomForestClassifier(n_estimators=item, random_state=0), X, y, cv=10) for item in param]
rf_train_acc = [sum(item==y)/len(y) for item in allyp] #train accuracy
i_best = np.argmax(rf_train_acc)
best_param = param[i_best]
best_acc = rf_train_acc[i_best]
print("Train Accuracy = " + str(best_acc) + ", n_estimators = "+ str(best_param))
```

```python
model = RandomForestClassifier(n_estimators=100, random_state=0).fit(X_train_norm, y_train)
y_predict = model.predict(X_test_norm)
rf_test_acc = sum(y_test == y_predict)/len(y_test) #test accuracy
print("Test Accuracy = " + str(rf_test_acc))
```
